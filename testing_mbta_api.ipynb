```python
with open("data/mbta_routes.json", "r") as f:
        data = json.load(f)
list_of_route_ids = []

for route in data["data"]:
    list_of_route_ids.append(route["id"])
    
# print(list_of_route_ids)
temp_route_ids = list_of_route_ids[0:5]
print(temp_route_ids)
```

```python
for route in list_of_route_ids:

    url = "https://api-v3.mbta.com/shapes"
    params = {
        # "include": "routes"
        # "sort": "short_name"
        "filter[route]": route
    }
    headers = {
        "x-api-key": API_KEY
    }

    resp = requests.get(url, params=params, headers=headers)
    data = resp.json()

    os.makedirs("data", exist_ok=True)
    with open(f"data/shapes/mbta_shapes_{route}.json", "w") as f:
        json.dump(data, f, indent=2)
    print("done")
```
